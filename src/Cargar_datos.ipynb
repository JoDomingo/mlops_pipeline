# Carga de datos

Este notebook realiza la carga inicial del dataset utilizado en el Proyecto Integrador del Módulo 5.

Para este Proyecto Integrador se utiliza el archivo `Base_de_datos.csv`, como base de datos.

El objetivo de este notebook es cargar el dataset y realizar verificaciones que confirmen que los datos fueron leídos correctamente antes de continuar con el análisis exploratorio.

```python
from pathlib import Path

import pandas as pd
```

```python
Path.cwd()
```

```python
ruta_datos = Path.cwd().parent / "Base_de_datos.csv"

print("Ruta del dataset:")
print(ruta_datos)

print("\¿El archivo existe?")
print(ruta_datos.exists())
```

```python
df = pd.read_csv(ruta_datos, sep=";")
```

```python
print(f"Filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")
```

```python
df.head()
```

```python
print("Columnas del dataset:")
for columna in df.columns:
    print(f"- {columna}")
```
